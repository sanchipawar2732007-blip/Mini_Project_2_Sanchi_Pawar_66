```python

import random

# DSA MINI PROJECT
# MAZE GENERATION USING DFS

ROWS = 10
COLS = 20

def generate_maze(rows, cols):
    # Create a grid filled with walls
    height = 2 * rows + 1
    width = 2 * cols + 1

    maze = [["#" for _ in range(width)] for _ in range(height)]

    # Track visited cells
    visited = set()

    # Start from the top-left cell
    start = (0, 0)
    stack = [start]
    visited.add(start)

    # Open the starting cell
    maze[1][1] = " "

    # DFS using a stack
    while stack:
        current_row, current_col = stack[-1]

        # Possible directions: up, down, left, right
        directions = [
            (-1, 0),
            (1, 0),
            (0, -1),
            (0, 1)
        ]

        random.shuffle(directions)
        found_neighbor = False

        for dr, dc in directions:
            new_row = current_row + dr
            new_col = current_col + dc

            # Check boundaries and visited status
            if (
                0 <= new_row < rows
                and 0 <= new_col < cols
                and (new_row, new_col) not in visited
            ):
                # Remove wall between current and new cell
                wall_row = 2 * current_row + 1 + dr
                wall_col = 2 * current_col + 1 + dc

                maze[wall_row][wall_col] = " "
                maze[2 * new_row + 1][2 * new_col + 1] = " "

                # Visit new cell
                visited.add((new_row, new_col))
                stack.append((new_row, new_col))
                found_neighbor = True
                break

        # Backtrack when no unvisited neighbor exists
        if not found_neighbor:
            stack.pop()

    # Mark start and finish
    maze[1][1] = "S"
    maze[2 * rows - 1][2 * cols - 1] = "E"

    return maze


def display_maze(maze):
    print("=" * 45)
    print("       MAZE GENERATION USING DFS")
    print("=" * 45)
    print()

    for row in maze:
        print("".join(row))

    print()
    print("S = Start")
    print("E = Exit")
    print("# = Wall")
    print("  = Path")
    print("\nMaze generated successfully!")


# Generate and display the maze
maze = generate_maze(ROWS, COLS)
display_maze(maze)
```
